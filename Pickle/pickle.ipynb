```python
import pickle

# pickle is used to serialize and deserialize a Python object structure, it can contain any type of object, even functions and classes

list1 = [
    "Toyota", "Honda", "Ford", "Chevrolet", "Nissan", "BMW", "Mercedes-Benz", "Volkswagen", 
    "Audi", "Hyundai", "Kia", "Lexus", "Subaru", "Mazda", "Dodge", "Jeep", "Chrysler", 
    "Buick", "Cadillac", "GMC", "Volvo", "Jaguar", "Land Rover", "Porsche", "Ferrari", 
    "Lamborghini", "Maserati", "Bentley", "Rolls-Royce", "Aston Martin", "Infiniti", 
    "Acura", "Lincoln", "Mitsubishi", "Mini", "Fiat", "Alfa Romeo", "Tesla", "Genesis", 
    "Suzuki", "Peugeot", "Renault", "Citroen", "Saab", "Skoda", "Seat", "Opel", "Isuzu", 
    "Daihatsu"
]

file = "mycar.pkl" # file name
filebject = open(file, "wb")  # open file in write binary mode
pickle.dump(list1, filebject)
filebject.close()
# the file is created in the same directory as the script

```

```python
# to read the file
file = "mycar.pkl" # file name
filebject = open(file, "rb")  # open file in read mode
mycar = pickle.load(filebject) # load the file
print(mycar)
filebject.close()
```
