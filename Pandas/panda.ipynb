## adding and removing rows and columns

```python
# Adding an new column
# creating a random array
arr = []
arr2 = []
for i in range(0,50):
    n1 = np.random.randint(5,15)
    arr.append(n1)
    n2 = np.random.randint(5,15)
    arr2.append(n2)

states["Price"] = arr2
states["coffe_price"] = arr 

states.head()
```

```python
# dropping a column
states.drop(columns = ["HS_Grad"], inplace = True)
states.head()
```

```python
# we can use copy function to copy one dataset into another
states_new = states.copy()
states_new["Zero_column"] = 0
states_new.head()
```

```python
# some calculations
states_new.drop(columns = ["Zero_column"], inplace = True)
states_new["final_price"] = states_new["Price"] * states_new["coffe_price"]
states_new.head()
```

```python
# Rename a column
states_new.rename(columns = {"State":'state', "Population":"Amount of People"}, inplace = True)
states_new.head(5)
```

```python
# convering date column to datetime data type
# removing excess columns
states_new.drop(columns = ["Murder", "Illiteracy", "Price", "coffe_price", "final_price"], inplace=True)
# creating date column
states_new["date"] = "1964-12-24"
states_new.info()
```

```python
# converting object date to real date
states_new["date"] = pd.to_datetime(states_new['date'])
states_new.info()
```

```python
# now we can separate the year, month and date
# all functions of datetime will be use in .dt.func() format
states_new["born_year"] = states_new["date"].dt.year
states_new[["state", "born_year"]].head(5)  # another simple of of displaying or selecting sepcific columns
```

```python
# saving data to a csv file
states_new.to_csv("datasets/states_new.csv", index = False)
# don't use .csv extension if you just want to export a simple file
```

```python
# applying lambda functions and adding it to a new column
states.drop(columns = ["coffe_price", "Price"], inplace = True)
states.head()
```

```python
states["remarks"] = states["Life_Exp"].apply(lambda x: "not good" if x < 70 else ("average" if x < 72 else "Good"))
states.head()
```

## Handling Null values

```python
# Creating Null values
states.loc[0:2, 'Life_Exp'] = np.nan
states.head()
```

```python
# finding out how many null values are there
states.isnull().sum()
```

```python
# # simple fillna
# states.fillna(100, inplace=True)
# states.head()
```

```python
# filling with mean
states.fillna(states['Life_Exp'].mean(), inplace=True)
states.head()
```

```python
# we can also use .interpolate(), or mode in place of .mean(), it will fill the value according to a pattern
# or we can drop the null rows
states.dropna(inplace=True)

# We can also drop rows according to specific columns
states.dropna(subset=["Life_Exp"], inplace=True) # drop the row if the life_exp in that is null

# to get the na value of a specific column
states[states["Life_Exp"].isna()]  # isnull == isna  notnull == notna
```
