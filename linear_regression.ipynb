```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# importing data
dataset = pd.read_csv("datasets/cleaned_data.csv")
dataset.head()
```

```python
# dropping the columns
dataset.drop(columns = ["emp_id", "duplicated", "department"], inplace = True)
dataset.head()
```

```python
x = dataset[["age","experience","en_department"]]
y = dataset["salary"]
```

```python
# Train test split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=42, test_size = 0.2)
```

```python
sns.scatterplot(x = "experience", y = "salary", data = dataset)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression as lr
lr = lr()
lr.fit(x_train, y_train)
```

```python
# testing the data
lr.predict([[42,10,3]])
```

```python
lr.score(x_test, y_test)*100
```

```python
# checking value
lr.coef_
```
